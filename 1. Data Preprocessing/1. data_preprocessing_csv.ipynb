### 3. Remove URL

```python
def remove_url(text:str):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

### 4. Remove Punctuation

```python
import string, time
string.punctuation
```

```python
exclude = string.punctuation
exclude
```

```python
def remove_punctuation(text:str):
    for char in exclude:
        text = text.replace(char, '')
    return text
```

```python
import time

text = "text. With punctuation?#%#$@^@RT$YYTgdgvfbhufeuhvbevhev  cvhufewj chvrchrebvrrv vchuevchre bchrbbcvcvcvvhedcve"

start = time.time()
print(remove_punctuation(text))
time2 = time.time() - start

print(time2 * 50000)
```

##### The above remove punctuation function is taking more time so it's better to use below function to remove the punctuations

```python
def remove_punctuation1(text:str):
    return text.translate(str.maketrans('', '', exclude))
```

```python
start = time.time()
print(remove_punctuation1(text))
time2 = time.time() - start
print(time2 * 50000)
```

### 5. Chat_conversation handling

```python
chat_words = {
    "AFAIK": "As Far As I Know",
    "AFK": "Away From Keyboard",
    "ASAP": "As Soon As Possible",
    "FYI": "For Your Information",
    "ASAP": "As Soon As Possible",
    "BRB": "Be Right Back",
    "BTW": "By The Way",
    "OMG": "Oh My God",
    "IMO": "In My Opinion",
    "LOL": "Laugh Out Loud",
    "TTYL": "Talk To You Later",
    "GTG": "Got To Go",
    "TTYT": "Talk To You Tomorrow",
    "IDK": "I Don't Know",
    "TMI": "Too Much Information",
    "IMHO": "In My Humble Opinion",
    "ICYMI": "In Case You Missed It",
    "AFAIK": "As Far As I Know",
    "BTW": "By The Way",
    "FAQ": "Frequently Asked Questions",
    "TGIF": "Thank God It's Friday",
    "FYA": "For Your Action",
    "ICYMI": "In Case You Missed It"
}

chat_words
```

```python
def chat_conversation(text:str):
    next_text = []
    for word in text.split():
        # print(word)
        if word.upper() in chat_words:
            next_text.append(chat_words.get(word.upper()))
            print(next_text)
        else:
            next_text.append(word)
            # print(next_text)
    return " ".join(next_text)

chat_conversation("Do this work ASAP")
```

```python
def remove_stopwords(text:str):
    result_text = []
    for word in text.split():
        if word in stopwords.words('english'):
            continue
        else:
            result_text.append(word)
    return " ".join(result_text)

remove_stopwords(text)
```

### 8. Remove emojis

```python
import re

def remove_emoji(text:str):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

remove_emoji(text)
```

```python
remove_emoji("Loved the movie. It was 😘😘")
```

### 10. Stemmer

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()

def stem_words(text):
    return " ".join(ps.stem(word) for word in text.split())

sample = "walk walks walking walked"
stem_words(sample)
```

### 11. Lemmatization

```python
import nltk
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word","Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
```

NOTE: Stemming & lamatization are same to retrieve root words but lamatization is worked good. Lamatization is slow & stemming is fast
