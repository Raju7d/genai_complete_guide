# HuggingFace
1. Huggingface is a platform, which is used to download all the opensource llm models, gen ai projects, nlp projects and computer vision projects
2. HF provides a **Pipeline** with help of this you can easily you can do any kind of taks
3. as of today the hugging face is having **21,50,467** models
4. to work with these models, you just need to use huggingface python library

### Main concepts of Huggingface (transformers) -> transformers is python hf library
    1. pipeline
    2. NLP taks
    3. Tokenization
    4. datasets and spaces (hf infra/ hf machine which is having good memory but we have to pay amount)
    5. fine tuning LLM using HF platform
    6. Project implementation (text summ, language trns, research paper summarization)

    Opensource LLM ==> HuggingFace
    Paid LLMs ==> Openai, Google, Amazon Bedrock
